##  Load the NER pipeline

```python
import torch
from transformers import AutoTokenizer, AutoModelForTokenClassification, pipeline

device = 0 if torch.cuda.is_available() else -1 # GPU/CPU

# dslim BERT based model for NER
tokenizer = AutoTokenizer.from_pretrained("dslim/bert-base-NER")
model = AutoModelForTokenClassification.from_pretrained("dslim/bert-base-NER")

ner_pipeline = pipeline(
    "ner",
    model=model,
    tokenizer=tokenizer,
    aggregation_strategy="max", # changed from "simple" to "max"
    device=device
)

print("NER pipeline loaded, using device:", "GPU" if device == 0 else "CPU")
```

## Entity Extraction on train and test





```python
from tqdm import tqdm #tqdm to show the progress bar

# func to run NER
def extract_entities(text):
    if not isinstance(text, str) or not text.strip(): # skip if not a valid text
        return []
    try:
        results = ner_pipeline(text[:512]) # fixed max input length 512 characters
        return [(r['word'], r['entity_group'], float(r['score'])) for r in results]
    except Exception as e:
        return []

tqdm.pandas()

train_df['entities'] = train_df['content'].progress_apply(extract_entities)
test_df['entities']  = test_df['content'].progress_apply(extract_entities)

print(train_df[['content', 'entities']].head(3))
```

## Preparing Entities for KG Linking

unique entity strings across train + test, Filter short characters, non-alphabetic characters, common english words), Count how many articles each entity appears in

```python
import re # used for pattern matching
from collections import Counter

def is_valid_entity(text):
    text = text.strip() #Remove spaces in a text

    if len(text) < 2: # Short to classify meaninful entity
        return False
    if not re.search(r'[a-zA-Z]', text):  # r=raw string , # must contain at least one letter
        return False
    if text.lower() in ['the', 'a', 'an', 'and', 'of', 'in', 'on']:  # Reject common english words
        return False
    return True

# Collect all unique entity strings
all_entity_texts = set()  # set() removes duplicate
for ents in train_df['entities']:
    for e in ents:
        all_entity_texts.add(e[0])
for ents in test_df['entities']:
    for e in ents:
        all_entity_texts.add(e[0])

cleaned_entities = set(t.strip() for t in all_entity_texts if is_valid_entity(t))
print(f"Raw unique entities: {len(all_entity_texts)} -> Cleaned: {len(cleaned_entities)}")
```

```python
# Frequency count
entity_counter = Counter()

# For train
for ents in train_df['entities']:
    for e in ents:
        if e[0].strip() in cleaned_entities:
            entity_counter[e[0].strip()] += 1
# For Test
for ents in test_df['entities']:
    for e in ents:
        if e[0].strip() in cleaned_entities:
            entity_counter[e[0].strip()] += 1

print(f"Total unique entities: {len(entity_counter)}")
print(f"Entities appearing only once: {sum(1 for c in entity_counter.values() if c == 1)}")
print(f"Entities appearing >=2 times:  {sum(1 for c in entity_counter.values() if c >= 2)}")
print(f"Entities appearing >=5 times:  {sum(1 for c in entity_counter.values() if c >= 5)}")

print("\nTop 20 most frequent entities:")
for ent, count in entity_counter.most_common(20):
    print(f"  {ent}: {count}")
```

## Linking Remaining Entities

```python
remaining = [e for e in entities_to_link if e not in entity_cache] # Remaining meaning subtracting the good ones
print(f"Remaining to link: {len(remaining)}")

consecutive_429s = 0

for i, entity_text in enumerate(tqdm(remaining)):
    success = False
    # Trying 3 times per entity before failing
    for attempt in range(3):
        try:
            resp = session.get(WIKIDATA_API, params={
                "action": "wbsearchentities", "search": entity_text,
                "language": "en", "format": "json", "limit": 1
            }, timeout=10)

            if resp.status_code == 429: # 429 = too many request
                consecutive_429s += 1
                wait = int(resp.headers.get("Retry-After", 5)) # 5 seconds to wait
                time.sleep(wait)
                continue

            consecutive_429s = 0
            data = resp.json()

            # If matched save the top result's ID, label, and description
            if data.get("search"):
                top = data["search"][0]
                entity_cache[entity_text] = {
                    "qid": top["id"],
                    "label": top.get("label", entity_text),
                    "description": top.get("description", "")
                }
            else:
                entity_cache[entity_text] = None # No matching entity
            success = True
            break
        except Exception:
            time.sleep(2) # If exception occured wait 2 sec and try agian

    if not success:
        entity_cache[entity_text] = None

    time.sleep(1.0)  # 1 request per sec for wikidata

    # If >=5 errors then stop
    if consecutive_429s >= 5:
        print("Too many request, fail, Stop")
        save_cache()
        break

    # To save results after evry 200 entities to drive
    if i % 200 == 0 and i > 0:
        save_cache()

save_cache() # Save when completes
print(f"Done for now. Cache has {len(entity_cache)} entities.")
good = sum(1 for v in entity_cache.values() if v is not None)
print(f"Success rate: {good/len(entity_cache)*100:.1f}%")
```

```python
import spacy
import spacy_entity_linker
import re

nlp = spacy.load("en_core_web_md")
nlp.add_pipe("entityLinker", last=True)

# stopword list to be excluded from spacy
STOPWORDS = nlp.Defaults.stop_words
print("Stop words removed")
```

## Entity cleaning and final semantic Linking

```python
def clean_entity_text(text): # punctuations, double quotes
    text = re.sub(r'^[\s"\'"''""„‚]+|[\s"\'"''""„‚]+$', '', text)
    return text.strip()

def is_valid_entity(text): # Delete short, no letters
    text = clean_entity_text(text)
    if len(text) < 2:
        return False
    if not re.search(r'[a-zA-Z]', text):
        return False
    if text.lower() in STOPWORDS:
        return False
    return True

def link_article_entities(article_text, ner_entities_with_labels, wikidata_cache): # Article linking
    raw_texts = list(set(
        clean_entity_text(e[0]) for e in ner_entities_with_labels
        if is_valid_entity(e[0])
    ))
    label_lookup = {clean_entity_text(e[0]): e[1] for e in ner_entities_with_labels}

    doc = nlp(article_text[:1000])
    context_links = {}
    for ent in doc._.linkedEntities:
        key = str(ent.get_span()).lower().strip()
        context_links[key] = {
            "qid": f"Q{ent.get_id()}",
            "label": ent.get_label(),
            "description": ent.get_description()
        }

    PERSON = {'PER', 'PERSON'}
    LOCATION_WORDS = ['city', 'town', 'village', 'county', 'municipality', 'commune',
                       'country', 'state', 'province', 'district']

    final = {}
    for ent_text in raw_texts:
        key = ent_text.lower()
        ner_label = label_lookup.get(ent_text, 'MISC')

        if key in context_links:
            desc = context_links[key]['description'].lower()
            is_location_match = any(w in desc for w in LOCATION_WORDS)

            if ner_label in PERSON and is_location_match:
                pass
            else:
                final[ent_text] = {**context_links[key], "source": "spacy_contextual"}
                continue

        if ent_text in wikidata_cache and wikidata_cache[ent_text] is not None:
            final[ent_text] = {**wikidata_cache[ent_text], "source": "wikidata_cache"}

    return final
```

## Wikidata Relationship (To check relationships)

```python
import time

def fetch_entity_relations_batch(qids): # search from Wikidata about entities
    params = { # wikidatas format
        "action": "wbgetentities",
        "ids": "|".join(qids),
        "props": "claims",
        "format": "json"
    }
    try:
        resp = session.get(WIKIDATA_API, params=params, timeout=15)
        if resp.status_code == 429:
            wait = int(resp.headers.get("Retry-After", 5))
            time.sleep(wait)
            return None
        return resp.json()
    except Exception:
        return None

def extract_all_entity_relations(claims): # Entity to Entity relationships
    relations = []
    for prop_id, claim_list in claims.items():
        for claim in claim_list:
            try:
                datavalue = claim["mainsnak"]["datavalue"]
                if datavalue["type"] == "wikibase-entityid":
                    target_qid = datavalue["value"]["id"]
                    relations.append((prop_id, target_qid))
            except (KeyError, TypeError):
                continue
    return relations
```

knowledge graph structure

```python
from tqdm import tqdm
import pickle, shutil, os

RELATIONS_CACHE_PATH = '/content/MIMoE-FND/kg_data/entity_relations.pkl'
DRIVE_RELATIONS_PATH = '/content/drive/MyDrive/Capstone_Dataset/kg_data/entity_relations.pkl'

def save_relations_cache():
    os.makedirs('/content/MIMoE-FND/kg_data', exist_ok=True)
    with open(RELATIONS_CACHE_PATH, 'wb') as f:
        pickle.dump(entity_relations, f)
    os.makedirs('/content/drive/MyDrive/Capstone_Dataset/kg_data', exist_ok=True)
    shutil.copy(RELATIONS_CACHE_PATH, DRIVE_RELATIONS_PATH)

remaining_qids = [q for q in all_qids if q not in entity_relations]
print(f"Remaining QIDs to fetch: {len(remaining_qids)}")

batch_size = 50
batches = [remaining_qids[i:i+batch_size] for i in range(0, len(remaining_qids), batch_size)]

for i, batch in enumerate(tqdm(batches)):
    data = fetch_entity_relations_batch(batch)
    if data is None:
        continue
    entities_data = data.get("entities", {})
    for qid, edata in entities_data.items():
        claims = edata.get("claims", {})
        entity_relations[qid] = extract_all_entity_relations(claims)
    time.sleep(1.0)
    if i % 20 == 0 and i > 0:
        save_relations_cache()

save_relations_cache()
print(f"Done. Fetched relations for {len(entity_relations)} entities.")
```

```python
#Run
from torch_geometric.data import Data

KG_DIM = entity_embeddings.shape[1]

def build_article_graph(entities_dict):
    qids, feats = [], []
    for _, data in entities_dict.items():
        qid = data.get('qid')
        if qid and qid in entity_to_id:
            qids.append(qid)
            feats.append(entity_embeddings[entity_to_id[qid]])

    if not qids:
        return None

    idx = {q: i for i, q in enumerate(qids)}
    x = torch.stack(feats)

    edges = []
    local_set = set(qids)
    for q in qids:
        for _, tail in entity_relations.get(q, []):
            if tail in local_set:
                edges.append([idx[q], idx[tail]])

    edge_index = (torch.tensor(edges, dtype=torch.long).t().contiguous()
                  if edges else torch.empty((2, 0), dtype=torch.long))

    return Data(x=x, edge_index=edge_index)
```

```python
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import GATConv, global_mean_pool

class ArticleGNN(nn.Module):
    def __init__(self, in_dim, hidden_dim=128, out_dim=128, heads=4):
        super().__init__()
        self.conv1 = GATConv(in_dim, hidden_dim, heads=heads, concat=True)
        self.conv2 = GATConv(hidden_dim * heads, out_dim, heads=1, concat=False)

    def forward(self, x, edge_index, batch):
        x = F.elu(self.conv1(x, edge_index))
        x = self.conv2(x, edge_index)
        return global_mean_pool(x, batch)
```

```python
#Run
from data.FakeNet_dataset import FakeNet_dataset
from torch_geometric.data import Data, Batch
from transformers import BertTokenizer, CLIPProcessor

class FakeNetKGDataset(FakeNet_dataset):
    def __init__(self, article_graphs, content_to_article_id, kg_dim, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.article_graphs = article_graphs
        self.content_to_article_id = content_to_article_id
        self.kg_dim = kg_dim
        self._empty_graph = Data(x=torch.zeros((1, kg_dim)), edge_index=torch.empty((2, 0), dtype=torch.long))

    def __getitem__(self, index):
        (content, img_GT, img_GT2, label, category), GT_path = super().__getitem__(index)
        article_id = self.content_to_article_id.get(content)
        graph = self.article_graphs.get(article_id, self._empty_graph) if article_id else self._empty_graph
        return (content, img_GT, img_GT2, label, category), GT_path, graph

word_token_length = 197
token_uncased = BertTokenizer.from_pretrained("bert-base-uncased")
clip_processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch16")

def collate_fn_english_kg(data):
    sents = [i[0][0] for i in data]
    image = [i[0][1] for i in data]
    image_aug = [i[0][2] for i in data]
    labels = [i[0][3] for i in data]
    category = [0 for i in data]
    GT_path = [i[1] for i in data]
    graphs = [i[2] for i in data]

    token_data = token_uncased(text=sents, truncation=True, padding="max_length",
                                max_length=word_token_length, return_tensors="pt", return_length=True)
    clip_inputs = clip_processor(text=sents, images=image, truncation=True, padding="max_length",
                                  max_length=77, return_tensors="pt", return_length=True)

    input_ids = token_data["input_ids"]
    attention_mask = token_data["attention_mask"]
    token_type_ids = token_data["token_type_ids"]
    image = torch.stack(image)
    image_aug = torch.stack(image_aug)
    labels = torch.LongTensor(labels)
    category = torch.LongTensor(category)
    graph_batch = Batch.from_data_list(graphs)

    return (input_ids, attention_mask, token_type_ids), (image, image_aug, labels, category, sents), clip_inputs, GT_path, graph_batch
```

## Baseline Fusion
KG/GNN is added at last step, merging after the baselines finished output, before the final classifier makes the prediction.

```python
class KGFusionModel(nn.Module):
    def __init__(self, baseline_model, gnn, gnn_out_dim=128):
        super().__init__()
        self.baseline_model = baseline_model
        self.gnn = gnn
        baseline_fused_dim = 64
        self.kg_classifier = nn.Linear(baseline_fused_dim + gnn_out_dim, 1)

        self._captured_fused_feature = None
        self.baseline_model.mix_classifier.register_forward_pre_hook(self._capture_pre_hook)

    def _capture_pre_hook(self, module, inputs):
        self._captured_fused_feature = inputs[0]

    def forward(self, input_ids, attention_mask, token_type_ids, image, clip_inputs, graph_batch):
        _, image_only_output, text_only_output, interaction_loss = self.baseline_model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            token_type_ids=token_type_ids,
            image=image,
            clip_inputs=clip_inputs,
        )

        # Combined here

        fused_feature = self._captured_fused_feature       # (batch, 64)
        gnn_embed = self.gnn(graph_batch.x, graph_batch.edge_index, graph_batch.batch)  # (batch, 128)

        combined = torch.cat([fused_feature, gnn_embed], dim=1)  # (batch, 192)
        kg_mix_output = self.kg_classifier(combined)

        return kg_mix_output, image_only_output, text_only_output, interaction_loss
```

Training

```python
from sklearn.metrics import precision_recall_fscore_support
from tqdm import tqdm


for epoch in range(NUM_EPOCHS):
    kg_model_exp1.train()
    train_loss_total = 0
    for batch in tqdm(train_loader, desc=f"Epoch {epoch+1}/{NUM_EPOCHS} [train]"):
        texts, others, clip_inputs_b, GT_path, graph_batch = batch
        input_ids, attention_mask, token_type_ids = texts
        image, image_aug, labels, category, sents = others
        input_ids, attention_mask, token_type_ids, image, labels = (
            input_ids.cuda(), attention_mask.cuda(), token_type_ids.cuda(),
            image.cuda(), labels.float().unsqueeze(1).cuda()
        )
        clip_inputs_b = clip_inputs_b.to("cuda")
        graph_batch = graph_batch.cuda()

        optimizer.zero_grad()
        mix_output, _, _, _ = kg_model_exp1(input_ids, attention_mask, token_type_ids, image, clip_inputs_b, graph_batch)
        loss = loss_fn(mix_output, labels)
        loss.backward()
        optimizer.step()
        train_loss_total += loss.item()

    avg_train_loss = train_loss_total / len(train_loader)

    kg_model_exp1.eval()
    all_preds, all_labels = [], []
    with torch.no_grad():
        for batch in tqdm(validate_loader, desc=f"Epoch {epoch+1}/{NUM_EPOCHS} [val]"):
            texts, others, clip_inputs_b, GT_path, graph_batch = batch
            input_ids, attention_mask, token_type_ids = texts
            image, image_aug, labels, category, sents = others
            input_ids, attention_mask, token_type_ids, image = (
                input_ids.cuda(), attention_mask.cuda(), token_type_ids.cuda(), image.cuda()
            )
            clip_inputs_b = clip_inputs_b.to("cuda")
            graph_batch = graph_batch.cuda()

            mix_output, _, _, _ = kg_model_exp1(input_ids, attention_mask, token_type_ids, image, clip_inputs_b, graph_batch)
            preds = torch.sigmoid(mix_output).round().squeeze().cpu().numpy()
            all_preds.extend(preds.tolist())
            all_labels.extend(labels.numpy().tolist())

    precision, recall, f1, _ = precision_recall_fscore_support(all_labels, all_preds, average='binary', zero_division=0)
    acc = sum(p == l for p, l in zip(all_preds, all_labels)) / len(all_labels)

    saved_msgs = []
    if acc > best_val_acc:
        best_val_acc = acc
        save_checkpoint(kg_model_exp1, optimizer, epoch + 1, acc, ACC_CKPT_PATH)
        saved_msgs.append("best_acc")

    if f1 > best_fake_f1:
        best_fake_f1 = f1
        save_checkpoint(kg_model_exp1, optimizer, epoch + 1, f1, F1_CKPT_PATH)
        saved_msgs.append("best_f1")

    saved_str = f" -> {', '.join(saved_msgs)}" if saved_msgs else ""
    print(f"Epoch {epoch+1}: Val_Acc={acc:.3f} (best: {best_val_acc:.3f}), Fake_F1={f1:.3f} (best: {best_fake_f1:.3f}){saved_str}")
```

```python
import re, spacy

nlp = spacy.load("en_core_web_md")
STOPWORDS = nlp.Defaults.stop_words

def clean_entity_text(text):
    text = re.sub(r'^[\s"\'"''""„‚]+|[\s"\'"''""„‚]+$', '', text)
    return text.strip()

def is_valid_entity(text):
    text = clean_entity_text(text)
    if len(text) < 2 or not re.search(r'[a-zA-Z]', text) or text.lower() in STOPWORDS:
        return False
    return True
```

```python
import re

singleton_entities = [ent for ent, count in entity_counter.items() if count == 1]

def looks_like_noise(text):
    if len(text.split()) > 5:
        return True
    if re.search(r"\s['’]\s", text):
        return True
    if '…' in text or text.count('.') > 1:
        return True
    return False

singleton_entities_filtered = [e for e in singleton_entities if not looks_like_noise(e)]
print(f"Singletons before filter: {len(singleton_entities)}")
print(f"Singletons after filter:  {len(singleton_entities_filtered)}")
```

Link Filtered Singletons via Wikidata

```python
from concurrent.futures import ThreadPoolExecutor, as_completed
from tqdm import tqdm
import time

def link_one_entity(entity_text):
    for attempt in range(3):
        try:
            resp = session.get(WIKIDATA_API, params={
                "action": "wbsearchentities", "search": entity_text,
                "language": "en", "format": "json", "limit": 1
            }, timeout=10)
            if resp.status_code == 429:
                time.sleep(3)
                continue
            data = resp.json()
            if data.get("search"):
                top = data["search"][0]
                return entity_text, {"qid": top["id"], "label": top.get("label", entity_text), "description": top.get("description", "")}
            return entity_text, None
        except Exception:
            time.sleep(2)
    return entity_text, None

remaining_singletons = [e for e in singleton_entities_filtered if e not in entity_cache]
print(f"Still to link: {len(remaining_singletons)}")

with ThreadPoolExecutor(max_workers=5) as executor:
    futures = {executor.submit(link_one_entity, e): e for e in remaining_singletons}
    for i, future in enumerate(tqdm(as_completed(futures), total=len(remaining_singletons))):
        entity_text, result = future.result()
        entity_cache[entity_text] = result
        if i % 200 == 0 and i > 0:
            save_cache()

save_cache()
print("Singleton linking done.")
```

```python
class KGFusionModel(nn.Module):
    def __init__(self, baseline_model, gnn, gnn_out_dim=128):
        super().__init__()
        self.baseline_model = baseline_model
        self.gnn = gnn
        self.kg_classifier = nn.Sequential(nn.Linear(64 + gnn_out_dim, 64), nn.SiLU(), nn.Linear(64, 1))
        self._captured_fused_feature = None
        self.baseline_model.mix_classifier.register_forward_pre_hook(self._capture_pre_hook)

    def _capture_pre_hook(self, module, inputs):
        self._captured_fused_feature = inputs[0]

    def forward(self, input_ids, attention_mask, token_type_ids, image, clip_inputs, graph_batch):
        _, image_only_output, text_only_output, interaction_loss = self.baseline_model(
            input_ids=input_ids, attention_mask=attention_mask, token_type_ids=token_type_ids,
            image=image, clip_inputs=clip_inputs)
        fused_feature = F.layer_norm(self._captured_fused_feature, self._captured_fused_feature.shape[1:])
        gnn_embed = self.gnn(graph_batch.x, graph_batch.edge_index, graph_batch.batch)
        gnn_embed = F.layer_norm(gnn_embed, gnn_embed.shape[1:])
        combined = torch.cat([fused_feature, gnn_embed], dim=1)
        return self.kg_classifier(combined), image_only_output, text_only_output, interaction_loss

gnn_v6 = ArticleGNN(in_dim=KG_DIM_V2).cuda()
kg_model_v6 = KGFusionModel(baseline_model, gnn_v6, gnn_out_dim=128).cuda()

CKPT_DIR_V6 = '/content/drive/MyDrive/Capstone_Dataset/kg_data/checkpoints/kg_fusion_v6_expanded'
os.makedirs(CKPT_DIR_V6, exist_ok=True)
ACC_CKPT_PATH_V6 = f'{CKPT_DIR_V6}/best_acc_model.pt'
F1_CKPT_PATH_V6 = f'{CKPT_DIR_V6}/best_f1_model.pt'

def save_checkpoint(model, optimizer, epoch, metric_value, path):
    torch.save({'epoch': epoch, 'model_state_dict': model.state_dict(), 'optimizer_state_dict': optimizer.state_dict(), 'metric_value': metric_value}, path)

optimizer_v6 = torch.optim.Adam(kg_model_v6.parameters(), lr=1e-5)
loss_fn_v6 = nn.BCEWithLogitsLoss()
best_fake_f1_v6 = 0.0
best_val_acc_v6 = 0.0
NUM_EPOCHS = 20
print("Exp5 model built.")
```

train_one_epoch

```python
from tqdm import tqdm

def train_one_epoch(baseline_model, gnn_from_exp4, train_loader, optimizer, loss_fn, device):
    baseline_model.train()
    gnn_from_exp4.eval()
    running_loss = 0.0

    for batch in tqdm(train_loader, desc="Exp6 Training"):
        texts, others, clip_inputs_b, GT_path, graph_batch = batch
        input_ids, attention_mask, token_type_ids = texts
        image, image_aug, labels, category, sents = others
        input_ids, attention_mask, token_type_ids = input_ids.to(device), attention_mask.to(device), token_type_ids.to(device)
        image = image.to(device)
        labels = labels.float().unsqueeze(1).to(device)
        clip_inputs_b = clip_inputs_b.to(device)
        graph_batch = graph_batch.to(device)

        optimizer.zero_grad()
        with torch.no_grad():
            kg_embedding = gnn_from_exp4(graph_batch.x, graph_batch.edge_index, graph_batch.batch)

        final_output, image_output, text_output, interaction_loss = baseline_model(
            input_ids=input_ids, attention_mask=attention_mask, token_type_ids=token_type_ids,
            image=image, clip_inputs=clip_inputs_b, kg_embedding=kg_embedding,
        )

        cls_loss = loss_fn(final_output, labels)
        loss = cls_loss + 0.1 * interaction_loss
        loss.backward()
        optimizer.step()
        running_loss += loss.item()

    return running_loss / len(train_loader)
```

validate

```python
from sklearn.metrics import precision_recall_fscore_support

def validate(baseline_model, gnn_from_exp4, validate_loader, loss_fn, device):
    baseline_model.eval()
    gnn_from_exp4.eval()
    running_loss = 0.0
    all_preds, all_labels = [], []

    with torch.no_grad():
        for batch in tqdm(validate_loader, desc="Exp6 Validation"):
            texts, others, clip_inputs_b, GT_path, graph_batch = batch
            input_ids, attention_mask, token_type_ids = texts
            image, image_aug, labels, category, sents = others
            input_ids, attention_mask, token_type_ids = input_ids.to(device), attention_mask.to(device), token_type_ids.to(device)
            image = image.to(device)
            labels_tensor = labels.float().unsqueeze(1).to(device)
            clip_inputs_b = clip_inputs_b.to(device)
            graph_batch = graph_batch.to(device)

            kg_embedding = gnn_from_exp4(graph_batch.x, graph_batch.edge_index, graph_batch.batch)
            final_output, image_output, text_output, interaction_loss = baseline_model(
                input_ids=input_ids, attention_mask=attention_mask, token_type_ids=token_type_ids,
                image=image, clip_inputs=clip_inputs_b, kg_embedding=kg_embedding,
            )

            cls_loss = loss_fn(final_output, labels_tensor)
            loss = cls_loss + interaction_loss
            running_loss += loss.item()

            preds = (torch.sigmoid(final_output) > 0.5).float()
            all_preds.extend(preds.squeeze().cpu().numpy().tolist())
            all_labels.extend(labels.cpu().numpy().tolist())

    avg_loss = running_loss / len(validate_loader)
    precision, recall, f1, _ = precision_recall_fscore_support(all_labels, all_preds, average="binary", zero_division=0)
    acc = sum(p == l for p, l in zip(all_preds, all_labels)) / len(all_labels)
    return avg_loss, acc, precision, recall, f1
```

```python
import pandas as pd
from data.FakeNet_dataset import FakeNet_dataset
from transformers import BertTokenizer, CLIPProcessor
from torch.utils.data import DataLoader
from torch_geometric.data import Data, Batch

train_df = pd.read_excel('/content/dataset/AAAI_dataset/gossip_train.xlsx')
test_df = pd.read_excel('/content/dataset/AAAI_dataset/gossip_test.xlsx')
content_to_article_id = {}
for i, row in train_df.iterrows():
    content_to_article_id[row['content']] = f'train_{i}'
for i, row in test_df.iterrows():
    content_to_article_id[row['content']] = f'test_{i}'

class FakeNetKGDataset(FakeNet_dataset):
    def __init__(self, article_graphs, content_to_article_id, kg_dim, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.article_graphs = article_graphs
        self.content_to_article_id = content_to_article_id
        self.kg_dim = kg_dim
        self._empty_graph = Data(x=torch.zeros((1, kg_dim)), edge_index=torch.empty((2, 0), dtype=torch.long))

    def __getitem__(self, index):
        (content, img_GT, img_GT2, label, category), GT_path = super().__getitem__(index)
        article_id = self.content_to_article_id.get(content)
        graph = self.article_graphs.get(article_id, self._empty_graph) if article_id else self._empty_graph
        return (content, img_GT, img_GT2, label, category), GT_path, graph

word_token_length = 197
token_uncased = BertTokenizer.from_pretrained("bert-base-uncased")
clip_processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch16")

def collate_fn_english_kg(data):
    sents = [i[0][0] for i in data]
    image = [i[0][1] for i in data]
    image_aug = [i[0][2] for i in data]
    labels = [i[0][3] for i in data]
    category = [0 for i in data]
    GT_path = [i[1] for i in data]
    graphs = [i[2] for i in data]
    token_data = token_uncased(text=sents, truncation=True, padding="max_length", max_length=word_token_length, return_tensors="pt", return_length=True)
    clip_inputs = clip_processor(text=sents, images=image, truncation=True, padding="max_length", max_length=77, return_tensors="pt", return_length=True)
    return (token_data["input_ids"], token_data["attention_mask"], token_data["token_type_ids"]), (torch.stack(image), torch.stack(image_aug), torch.LongTensor(labels), torch.LongTensor(category), sents), clip_inputs, GT_path, Batch.from_data_list(graphs)

GT_size = 224
custom_batch_size = 24

train_kg_dataset = FakeNetKGDataset(article_graphs=article_graphs, content_to_article_id=content_to_article_id, kg_dim=KG_DIM, is_train=True, dataset="gossip", image_size=GT_size, data_augment=False, duplicate_fake_times=0)
validate_kg_dataset = FakeNetKGDataset(article_graphs=article_graphs, content_to_article_id=content_to_article_id, kg_dim=KG_DIM, is_train=False, dataset="gossip", image_size=GT_size)

train_loader = DataLoader(train_kg_dataset, batch_size=custom_batch_size, shuffle=True, collate_fn=collate_fn_english_kg, num_workers=4, drop_last=True, pin_memory=True)
validate_loader = DataLoader(validate_kg_dataset, batch_size=custom_batch_size, shuffle=False, collate_fn=collate_fn_english_kg, num_workers=4, drop_last=False, pin_memory=True)
print(f"Train: {len(train_kg_dataset)}, Validate: {len(validate_kg_dataset)}")
```
